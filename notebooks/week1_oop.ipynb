## Numpy
### Arrays
Lets start with some basics to declare multi-dimensional arays using the numpy package. Numpy arays are more efficient and convenient than nested lists.

```python
import numpy as np

ar1 = np.array([[1,2,3,4],[5,6,7,8]])
print(ar1)

print(f'\nElement an Position (1,2) ist: {ar1[0,1]} \n')

print(f'Anzahl an Dimensionen: {ar1.ndim}')
print(f'Anzahl an Elementen: {ar1.size}')
print(f'Form des Arrays (Zeilen, Spalten): {ar1.shape}')
print(f'Datentyp des Arrays: {ar1.dtype}')

ar1 = ar1.astype(np.int32)
print(f'Neuer Datentyp des Arrays: {ar1.dtype}')
```

Numpy functions to analyze or modify an array:
- **sum()**: Returns sum of all elements or columns/rows separately
- **min()**: Minimum value within array
- **max()**: Maximum value within array
- **mean()**: Mean value of array elements
- **median()**: Median value of array elemtns
- **minimum()**: Returns an array with minimum value in each position of array
- **maximum()**: Returns an array with maximum value in each position of array 

Ermittlung der Zeilen oder Spaltensummen eines Arrays

```python
spaltensummen = ar1.sum(axis=0)
print(f'Spaltensummen des Arrays ar1: {spaltensummen}\n')
for i, spaltensume in enumerate(spaltensummen):
    print(f'Spalte {i+1}: {spaltensume}')

zeilensummen = ar1.sum(axis=1)
print(f'\nZeilensummen des Arrays ar1: {zeilensummen}\n')
for i, zeilensumme in enumerate(zeilensummen):
    print(f'Zeile {i+1}: {zeilensumme}')
```

Kummulierte Summe

```python
print(f'Kummuliert gesamt:\n{np.cumsum(ar1)}')
print(f'Kummuliert Spaltenweise:\n{np.cumsum(ar1, axis=0)}')
print(f'Kummuliert Dimensionsweise:\n{np.cumsum(ar1, axis=1)}')
```

There are several predefined numpy functions to setup arrays of specific kind

```python
einser_array = np.ones((2,3), dtype=np.int32)
print(f'Eins-Array:\n{einser_array}\n')

nuller_array = np.zeros((2,3), dtype=np.int32)
print(f'Null-Array:\n{nuller_array}\n')

zufalls_array = np.random.rand(2,3)
print(f'Zufalls-Array:\n{zufalls_array}')
zufalls_array2 = np.random.random((2,3))
print(f'Zufalls-Array 2:\n{zufalls_array2}\n')

fixer_array = np.full((2,3), 8, dtype=np.int32)
print(f'Fixer-Array:\n{fixer_array}\n')

einheits_matrix = np.eye(4)
print(f'Einheits-Matrix:\n{einheits_matrix}')
```

### Sequences
Creating sequences is another fundamental point for operations research, e.g. defining variables or parameters with indices. There are basically two different methods to create sequences
- **arange(start,end,step)**: Creates a numpy array with elements ranging from *start* to *end* (if no start is determined, its start at 0 by default). The step size is manually determined by the *step* attribute.
- **linspace(start,end,numvalues)**: Creates a numpy array with **numvalues* elements, evenly distributed between *start* and *end*. The step size is thus automatically calculated by the function.

```python
seq1 = np.arange(0, 10, 3)
print(f'Sequence 1 (arange): {seq1}')

seq2 = np.linspace(0, 10, 5)
print(f'Sequence 2 (linspace): {seq2}')
```

### List comprehension
Another important thin is list comprehension to assign list variable efficiently.

```python
base = [1,2,3,4]
squares = [i**2 for i in base]
print(f'Squares using list comprehension: {squares}')

variable_names = [f'x_{i}' for i in base]
print(f'Variable names using list comprehension: {variable_names}')
```

### Main concepts
Matplotlib introduces three basic components:
- **figure**: Is the canvas where the data is plotted, i.e. the container arranging one or more graphs (plot)
- **plot**: Are the actual graphs which are included and displayed within the figure. It has a *title*, a *legend* and 2 or more *axis*.
- **axis**: Conform the limits of a plot

```python
y = 2 + np.random.rand(100)
x = np.arange(y.size)

plt.plot(x, y, 'o--', label = 'Random function')
plt.title('Random Data Plot')
plt.legend(loc = 'upper left')
```

```python
# y2 = np.array([i**2 for i in y])
y2 = y**2

plt.plot(x, y, 'o--', label = 'Random function')
plt.plot(x, y2, '*-.', label = 'Cubic function')
plt.title('Random Data Plot with Cubic Function')
plt.legend(loc = 'upper left')
```

### Formatting axes
Basic functions to format the axes are
- **xlim(left,right)**: Limits of x-axis provided as tuple of integer values
- **ylim(bottom,top)**: Limits of y-axis provided as tuple of integer values
- **xlabel(label)**, **ylabel(label)**: Label of x-axis and y-axis, using a string (also latex formula with $-signs possible)

### Formatting plot
- **plt.fill_between(x,y,y2=0,where=None,args)**: Fills the area between lines *y* (upper limit of area) and *y2* (lower limit of area), both versus *x*. If a scalar value is used, a horizontal line crossing the y-axis at the value provided is used as bottom limit of the area (default 0). The optional *where* argment is used to provide an expressionin *x* to limit the area in the horizontal axis. Further arguments such as *color* or *alpha* (transparency) can be found in the documentation.
- **axvline(x=0,ymin=0, ymax=1, args)**: Draws a horizontal line at the scalar position of *x*. The additonal *ymin* and *ymax* arguments define the lower and ppoer limits of the line.
