### Aufgabe 2 - Einfache Funktionen (9 Punkte)

#### a.)
#### (4 Punkte)

Erstellen Sie eine Funktion *analyse_verspaetung*, die eine Verspätung in Minuten als Parameter erhält und folgende Ausgaben erzeugt:

- Wenn die Verspätung 0 ist: "Der Zug ist pünktlich."
- Wenn die Verspätung zwischen 1 und 5 Minuten liegt: "Geringe Verspätung."
- Wenn die Verspätung zwischen 6 und 15 Minuten liegt: "Mäßige Verspätung."
- Wenn die Verspätung über 15 Minuten beträgt: "Erhebliche Verspätung!"

```python
### Bitte ergänzen Sie hier Ihren Programmcode. ###
def analyse_verspaetung(verspaetung):
    if verspaetung == 0:
        return 'Der Zug ist pünktlich.'
    elif verspaetung <= 5:
        return 'Geringe Verspätung.'
    elif verspaetung <= 15:
        return 'Mäßige Verspätung.'
    else:
        return 'Erhebliche Verspätung.'


### Führen Sie anschließend nachfolgenden Programmcode aus, um Ihre Lösung zu überprüfen. ###

analyse_verspaetung(10)
```

Erwarteter Output:

    Mäßige Verspätung.

#### b.)
#### (5 Punkte)

Schreiben Sie eine Funktion *statistik_verspaetungen*, die eine Liste von Verspätungswerten erhält und folgende Werte berechnet:

- Durchschnittliche Verspätung (gerundet auf eine Nachkommastelle).
- Anteil ausreichend pünktlicher Züge (Verspätung < 6 Minuten) in Prozent.
- Maximale Verspätung in der Liste.

Die Funktion soll die Werte anschließend als formatierte Ausgabe ausgeben.

```python
### Bitte ergänzen Sie hier Ihren Programmcode. ###
def statistik_verspaetungen(verspaetungen):
    avg_delay = sum(verspaetungen)/len(verspaetungen)
    punctual_quota = (len([v for v in verspaetungen if v < 6])/len(verspaetungen))*100
    # punctual_quota = sum(1 for v in verspaetungen if v < 6)/len(verspaetungen)*100    # alternatively with sum() and generator expression
    max_delay = max(verspaetungen)
    
    print(f'Durchschnittliche Verspätung: {avg_delay:.1f} Min')
    print(f'Anteil ausreichend pünktlicher Züge: {punctual_quota:.1f} %')
    print(f'Maximale Verspätung: {max_delay} Min')


### Führen Sie anschließend nachfolgenden Programmcode aus, um Ihre Lösung zu überprüfen. ###

statistik_verspaetungen([5, 10, 3, 0, 8, 1, 14, 7, 2, 2, 11, 4, 5, 9, 0, 25])
```

Erwarteter Output:

    Durchschnittliche Verspätung: 6.6 Minuten
    Anteil ausreichend pünktlicher Züge: 56.2 %
    Maximale Verspätung: 25 Minuten

### Aufgabe 3 - Einlesen einer Datei (8 Punkte)

In der beigefügten csv-Datei sind die Straßenbahnlinien der Dresdner Verkehrsbetriebe samt Länge in km und Fahrzeit in Minuten enhalten.

#### a.)
#### (4 Punkte)

Lesen Sie die beigefügte csv-Datei ein und speichern Sie die Linien in einem Dictionary *linien_info* ab. Die Liniennummer soll dabei der Schlüssel sein. Zu jeder Linie ergeben sich wiederum als Dictionary die Schlüssel-Wert-Paare für Länge (distance), Fahrzeit (duration) und berechneter Durchschnittsgeschwindigkeit in km/h (average_speed) an. Runden Sie auf 2 Nachkommastellen genau und lassen Sie sich das Dictionary ausgeben.

```python
import csv

### Bitte ergänzen Sie hier Ihren Programmcode. ###
linien_info = dict()

with open('../../data/Prescriptive_Homework/dvb.csv', mode='r', encoding='utf-8') as inputfile:
    data = csv.DictReader(inputfile, delimiter=';')

    for row in data:
        line = int(row['Linie'])
        distance = float(row['Länge (km)'])
        duration = int(row['Fahrzeit (Minuten)'])
        avg_speed = round(distance/(duration/60),1)

        linien_info[line] = {
            'distance': distance, 
            'duration': duration, 
            'average_speed': avg_speed
        }

### Führen Sie anschließend nachfolgenden Programmcode aus, um Ihre Lösung zu überprüfen. ###

print(linien_info)
```

Erwarteter Output:

    {1: {'distance': 15.2, 'duration': 47, 'average_speed': 19.4}, ... }

```python
### Ergänzen Sie hier Ihren Code, um die gewünschte Ausgabe zu erzielen. ###
with open('../../data/Prescriptive_Homework/tram_lines.txt', mode='w', encoding = 'utf-8') as outputfile:
    outputfile.write('Dresdner Straßenbahnlinien\n')
    outputfile.write('===========================\n')

    longest_duration = max(linien_info.items(), key=lambda x: x[1]['duration'])
    longest_line = max(linien_info.items(), key=lambda x: x[1]['distance'])
    highest_speed = max(linien_info.items(), key=lambda x: x[1]['average_speed'])

    
    outputfile.write(f'   Anzahl Linien: {len(linien_info)}\n')
    outputfile.write(f'   Linie mit längster Fahrtzeit: {longest_duration[0]} ({longest_duration[1]['duration']} Min)\n')
    outputfile.write(f'   Linie mit längster Strecke: {longest_line[0]} ({longest_line[1]["distance"]} km)\n')
    outputfile.write(f'   Schnellste Linie: {highest_speed[0]} ({highest_speed[1]['average_speed']} km/h)\n')
    outputfile.write('===========================\n\n')

    
    for linie, info in linien_info.items():
        outputfile.write(f'Linie {linie}:\n')
        outputfile.write(f'  - Länge: {info["distance"]} km\n')
        outputfile.write(f'  - Fahrzeit: {info["duration"]} Minuten\n')
        outputfile.write(f'  - Durchschnittsgeschwindigkeit: {info["average_speed"]} km/h\n\n')
```
