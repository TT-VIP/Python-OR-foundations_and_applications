#### a.) Die Fahrzeug-Klasse
#### (6 Punkte)

Erstellen Sie eine Klasse Vehicle, die als Parameter eine Id *id*, eine Marke *brand*, ein Modell *model* und eine maximale Geschwindigkeit *max_speed* entgegennimmt. Die Klasse soll eine Methode *accelerate(self, value)* enthalten, die die aktuelle Geschwindigkeit um value erhöht, jedoch nicht über die maximale Geschwindigkeit hinausgeht. Zusätzlich soll eine Methode *brake(self, value)* implementiert werden, die die aktuelle Geschwindigkeit um value reduziert, aber nicht unter 0 fällt.

Eine *__str__()-Methode* soll eine passende Beschreibung des Fahrzeugs ausgeben.

```python
### Bitte ergänzen Sie hier Ihren Programmcode ###
class Vehicle:
    def __init__(self, id, brand, model, max_speed):
        self.id = id
        self.brand = brand
        self.model = model
        self.max_speed = max_speed
        self.speed = 0

    @property
    def tempo(self):
        return self.speed
    
    def accelerate(self, value):
        if value < 0:
            return 'Nutzen Sie die Bremse mit der Methode brake() um die Geschwindigkeit zu verringern.'
        elif self.speed + value > self.max_speed:
            self.speed = self.max_speed
            return f'Die Maximalgeschwindigkeit von {self.max_speed} km/h wurde erreicht.'
        else:
            self.speed += value
            return f'Die Geschwindigkeit wurde um {value} auf {self.speed} km/h erhöht.'
        
    def brake(self, value):
        if value < 0:
            return 'Nutzen Sie die Methode accelerate() um die Geschwindigkeit zu erhöhen.'
        elif self.speed - value < 0:
            self.speed = 0
            return 'Das Fahrzeug steht jetzt.'
        else:
            self.speed -= value
            return f'Die Geschwindigkeit wurde um {value} auf {self.speed} km/h verringert.'
        
    def __str__(self):
        return f'class: Vehicle, brand: {self.brand}, model: {self.model}, max. Speed: {self.max_speed} km/h'
        
    

### Führen Sie anschließend nachfolgenden Programmcode aus, um Ihre Lösung zu überprüfen ###

my_vehicle = Vehicle(0, "BMW", "X5", 250)
print(my_vehicle)
```

Erstellen Sie eine Klasse *Warteschlange*. Bei Initialisierung von Objekten der Klasse soll optional eine Liste übergeben werden können. Wird keine Liste übergeben, so wird eine leere Liste initialisiert.

Definieren Sie zusätzlich die Methoden *add*, *remove* und *isEmpty*. Methode *add* fügt eine Element an das Ende der Liste an. 

Die Methode *remove* soll Elemente aus der Liste nach dem FIFO-Prinzip entnehmen. Achten Sie darauf, dass keine Elemente aus einer leeren Liste entnommen werden können.

Schließlich prüft *isEmpty*, ob in der Liste noch Elemente vorhanden sind.

**Tipp:**
<details>

FIFO steht für First-In-First-Out, d.h.: Elemente, die als erstes in eine Liste/ Wartschlange aufgenommen wurden, werden auch als erstes wieder entfernt.
</details>


```python
### Bitte ergänzen Sie hier Ihren Programmcode ###
class Warteschlange:
    def __init__(self, list=None):
        self.queue = [] if list is None else list
    
    def isEmpty(self):
        return len(self.queue) == 0
    
    def add(self, element):
        self.queue.append(element)

    def remove(self):
        if self.isEmpty():
            return None
        else:
            self.queue.pop(0)
    


### Führen Sie anschließend nachfolgenden Programmcode aus, um Ihre Lösung zu überprüfen ###

for l in [[], [1, 2, 3], [5]]:
    myQueue = Warteschlange(l)
    print(myQueue.queue)
    myQueue.remove()
    print(myQueue.isEmpty())

```

#### b.) Performance-Vergleich
#### (7 Punkte)

Gegeben ist die Funktion *measureTime*, welche als Argumente eine Funktion(*func*), die notwendigen Funktionsargumente (*funcArg*) und die Anzahl an Testwiederholungen (*n*) entgegen nimmt. 

Die Funktion *measureTime* führt anschließend die übergebene Funktion *n*-mal aus und misst die Rechenzeit. Anschließend gibt sie den Mittelwert aller *n* Iterationen zurück. Zur Veranschlaulichung können Sie den nachfolgenden Code ausführen.

```python
import timeit ## Import eines Moduls zur Zeitmessung

def measureTime(func, funcArg, n): ## Funktionsdefinition
    t = timeit.Timer(lambda: func(funcArg)) ## Intialisierung eines Timer-Objekts mit den gegebenen Argumenten
    result = t.timeit(n) ## Ausführung der n-Durchläufe
    print(f"Execution time is {result / n} seconds") ## Ausgabe

# Eine Beispiel-Funktion, welche die Zahlen von 0 bis x summiert.
def addition(x):
    print('Addition:', sum(range(x)))

# Definition der Iterationsanzahl n
n = 5

# Ausführen des Tests
measureTime(addition, 100000, 5)
```
