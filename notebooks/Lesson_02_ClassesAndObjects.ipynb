### a.)

Erstellen Sie die Klasse mit den genannten Eigenschaften und führen Sie anschließend die unten stehenden Befehle aus.


```python
class Podcast:
    def __init__(self, episode, title, length, moderator, adverts = 0):
        self.episode = episode
        self.title = title
        self.length = length
        self.moderator = moderator
        self.adverts = adverts

    def display(self):
        print(f'''
              Podcast Folge Nr.: {self.episode} mit ModeratorIn: {self.moderator}\n
              Thema der Podcastfolge: {self.title}\n
              Länge der Podcastfolge: {self.length} Minuten\n
              Anzahl der Werbeblöcke: {self.adverts}
              ''')

newpodcast = Podcast(23, "Traveling Salesman Problems", 70, "O. Peratio")

newpodcast.display()
```

### b.)

Die Grundstruktur der Klasse steht. Nun sollen weitere Anpassungen vorgenommen werden:

1. Die Attribute des Objekts sollen nach Initialisierung nicht mehr von außen angesteuert werden können.
2. Neben der bestehenden Methode sollen noch weitere Methoden intergiert werden:
- Die Methode `setAdverts` fügt der Länge des Podcasts die 3-minütigen Werbeblöcke hinzu. Wie viele Werbeblöcke hinzugefügt werden, hängt von der mit gelieferten Anzahl ab. Die Anzahl der Werbeblöcke wird im entsprechenden Attribut gespeichert.
- Die Methode `getAdverts` gibt die aktuelle Anzahl an Werbeblöcken aus.
- Mit der Methode `cut` kann ein Podcast gekürzt werden. Wie viele Minuten herausgeschnitten werden, hängt wiederum vom übergebenen Parameter ab. Jedoch soll dabei ein Grenzwert von 30 Minuten nicht unterschritten werden. Bei Unterschreitung dieser Grenze soll eine entsprechende Meldung erscheinen.

Erweitern Sie die Klasse *Podcast* entsprechend der Angaben und führen Sie anschließend den entsprechenden Code aus.

```python
class Podcast:
    def __init__(self, episode, title, length, moderator, adverts = 0):
        self.__episode = episode
        self.__title = title
        self.__length = length
        self.__moderator = moderator
        self.__adverts = adverts

    def display(self):
        print(f'Podcast Folge Nr.: {self.__episode} mit ModeratorIn: {self.__moderator}\n'
              f'Thema der Podcastfolge: {self.__title}\n'
              f'Länge der Podcastfolge: {self.__length} Minuten\n'
              f'Anzahl der Werbeblöcke: {self.__adverts}\n')
        
    # @property
    # def advert_number(self):
    #     return self._adverts
    
    # @advert_number.setter
    # def setAdverts(self, number):
    #     if self._adverts + number <= 3:
    #         self._adverts += number
    #     else:
    #         print(f"Achtung: Maximal 3 Werbeblöcke erlaubt! Es können nur noch {3-self._adverts} Werbeblöcke hinzugefügt werden.")

    def setAdverts(self, number):
        if self.__adverts + number <= 3:
            self.__adverts += number
            self.__length += number*3   # each advert adds 3 minutes to the length
        else:
            print(f"Achtung: Maximal 3 Werbeblöcke erlaubt! Es können nur noch {3-self.__adverts} Werbeblöcke hinzugefügt werden.")

    def getAdverts(self):
        return self.__adverts
    
    def cut(self, minutes):
        if self.__length - minutes >= 30:
            self.__length -= minutes
        else:
            print(f'Achtung: Die Podcastfolge muss mindestens 30 Minuten lang sein! Es können nur noch {self.__length-30} Minuten\n'
                  f'         gekürzt werden.')

newpodcast = Podcast(23, "Traveling Salesman Problems", 70, "O. Peratio")

newpodcast.display()

newpodcast.setAdverts(2)
newpodcast.setAdverts(2)

newpodcast.cut(45)
newpodcast.cut(10)

newpodcast.display()
```

Als letzter Schritt soll nun eine Klasse *SpecialPodcast* eingeführt werden, die neben den bestehenden Eigenschaften auch noch einen *specialguest* beinhalten soll. Da die Bestätigung der Gäste jedoch immer relativ kurzfristig erfolgt, kann mittels der Methode `changeGuest` der *specialguest* nochmal verändert werden.

Bitte erstellen Sie die Klasse mittels Vererbung und nehmen Sie die zusätzliche Eigenschaft sowie die neue Methode mit in die Klasse auf. Führen Sie anschließend die unten stehenden Befehle aus.

```python
class SpecialPodcast(Podcast):
    def __init__(self, episode, title, length, moderator, guest, adverts = 0):
        super().__init__(episode, title, length, moderator, adverts)
        self.__guest = guest

    def changeGuest(self, new_guest):
        self.__guest = new_guest

    def display(self):
        print(f'Podcast Folge Nr.: {self._Podcast__episode} mit ModeratorIn: {self._Podcast__moderator}\n'
              f'Thema der Podcastfolge: {self._Podcast__title}\n'
              f'Länge der Podcastfolge: {self._Podcast__length} Minuten\n'
              f'Anzahl der Werbeblöcke: {self._Podcast__adverts}\n'
              f'Sondergast: {self.__guest}\n')

sp = SpecialPodcast(1, "Vehicle Routing Problem", 65, "M. Ipler", "Dr. Best")

sp.changeGuest("Dr. Secondbest")

sp.display()

sp.getAdverts()
sp.setAdverts(3)
sp.cut(-20)
sp.display()
```

Nach der ersten Besichtigung der Fertigung können Sie das Planungsproblem als klassisches Flow Shop Problem mit Rüstzeiten identifizieren:

Ein Set $N$ = {1, 2, . . . , $n$} aus $n$ Aufträgen soll sequenziell auf $m$ unterschiedlichen Stationen bearbeitet werden, wobei an jeder Station nur eine Maschine zur Verfügung steht. Die Reihenfolge der Arbeitsgänge ist für alle Aufträge identisch. Alle Maschinen sind durchgehend verfügbar und können jeweils einen Auftrag gleichzeitig bearbeiten. Die Puffermöglichkeit zwischen den Maschinen sind als unendlich anzunehmen. Bevor jedoch ein Auftrag auf einer Maschine bearbeitet werden kann, muss die Maschine vorbereitet bzw. gerüstet werden.

Als kundenorientierter Unternehmer möchte Dr. Best am liebsten alle KundInnen mit möglichst pünktlichen Lieferungen zufrieden stellen. Allerdings sind manche KundInnen wichtiger als andere, da diese eine Verzögerung der Lieferungen mit hohen Strafkosten belegt haben.

### b.)

Welche Klassen benötigen Sie in Python, um das Planungsproblem grundsätzlich zu beschreiben? Denken Sie daran die `__init__()`- und `__str__()`-Funktion zu definieren. Die Inputdaten aus der json-Datei sollten sich in den Klassen wieder finden.

Setzen Sie diese Klassen in Python um.

```python
class Machine:
    def __init__(self, machId):
        self.__machId = machId

    def __str__(self):
        return f'Maschine {self.__id}'
    
    @property
    def MachineId(self):
        return self.__machId
    

class Job:
    def __init__(self, jobId, setup_times, processing_times, due_date, tard_cost):
        self.__jobId = jobId
        self.__setup_times = setup_times
        self.__processing_times = processing_times
        self.__due_date = due_date
        self.__tard_cost = tard_cost

    def __str__(self):
        return (
            f'Job {self.__jobId} mit {len(self.__processing_times)} Arbeitsschritten.\n'
            f'Rüstzeiten {self.__setup_times} und Bearbeitungszeiten {self.__processing_times}, Fälligkeitsdatum {self.__due_date} und Verspätungskosten {self.__tard_cost}'
        )
    
    @property
    def JobId(self):
        return self.__jobId
    
    @property
    def DueDate(self):
        return self.__due_date
    
    @property
    def TardCost(self):
        return self.__tard_cost
    
    @property
    def ProcessingTimes(self):
        return self.__processing_times
    
    @property
    def SetupTimes(self):
        return self.__setup_times
    
    @property
    def Operations(self):
        return [(opId,processTime) for opId, processTime in enumerate(self.__processing_times)]
```
